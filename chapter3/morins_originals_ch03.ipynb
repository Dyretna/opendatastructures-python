```python
from collection import SLList, DLList, SEList
```

## Single Linked List

```python
sll = SLList()

sll.push("a")
sll.push("b")
sll.push("c")
sll.push("d")
sll.push("e")
sll.push("f")

sll.pop()
print(sll)
```

```python
print(sll.tail.next)

sll.add("x")
sll.add("y")
sll.add("z")

sll.remove()
print(sll)
```

## Double Linked List

```python
dll = DLList()

dll.add(0, "a")
dll.add(1, "b")
dll.add(2, "c")
dll.add(3, "d")
dll.add(3, "d2") # add before, as list.insert()
dll.add(3, "d3")

print(dll)
dll.remove(2)
print(dll)
```

# Space-Efficient Linked List

### Bounded deck

```python
from collection import BDeque

bd = BDeque(capacity=8)

bd.add_left("a")
bd.add_left("b")
bd.add_left("c")
print(" add_left ".center(30, "-") + "\n" + str(bd))

bd.add_right("d")
bd.add_right("e")
bd.add_right("f")
print(" add_right ".center(30, "-") + "\n" + str(bd))


bd.add(2, "insert 1")
bd.add(3, "insert 2")
print(" add inserting ".center(30, "-") + "\n" + str(bd))
```

```python
bd.remove(0)
bd.remove(0)
print(" remove 0 ".center(30, "-") + "\n" + str(bd))
```

```python
sel = SEList(block_size=3) # 3 + 1

sel.add(0, "a")
sel.add(0, "b")
sel.add(0, "c")
sel.add(0, "d")

sel.append("e")
sel.append("f")
sel.append("g")
sel.append("h")

sel.add(4, "S")
sel.add(5, "T")
sel.add(6, "U")
sel.add(7, "W")

# SEList __str__() corrects order with deques get method,
# and inserts empty slots with Nones (capacity - n)
print(sel)
```

### trigger spread

```python
sel.add(1, "x")

print(sel)
```

```python
print(sel.remove(2))
print(sel)
print(sel.remove(1))
print(sel)
print(sel.remove(0))
print(sel)
```
