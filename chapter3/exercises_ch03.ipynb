## Exercise 3.2. second_last()

Design and implement an SLList method, second last(),
that returns the second-last element of an SLList. Do this without using the member variable, n, that keeps track of the size of the list.

```python
# forward traversing with head.next

from collection import SLList

sll = SLList()
for i in range(5):
    sll.push(i)

print(sll)
print("\nSecond last: ", sll.second_last())

sll = SLList()
for i in range(5):
    sll.add(i, i)

print(sll)
print("\nSecond last: ", sll.second_last())
```

## Exercise 3.7. is_palindrome()

Write a DLList method is_palindrome() that returns true
if the list is a palindrome, i.e., the element at position i is equal to the
element at position n − i − 1 for all i ∈ {0, . . . , n − 1}. Your code should run
in O(n) time.

```python
word = "sallad dallas"

dll = DLList()
for i in range(len(word)):
    dll.add(i, word[i])

dll.is_palindrome()
```
