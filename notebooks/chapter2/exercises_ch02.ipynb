## Exercise 2.3. Treque() 

Design and implement a Treque (triple-ended queue). This is a List implementation in which get(i) and set(i, x) run in constant time and add(i, x) and remove(i) run in time O(1 + min{i, n − i, |n/2 − i|}) .

In other words, modifications are fast if they are near either end or near the middle of the list.

```python
from ods.arrays import Treque

# end operations does not balance the blocks,
# only with indexing add(i, x) and remove(i)

treq = Treque()
for i in range(9):
    treq.add_first(i)
print(treq)

treq = Treque()
for i in range(9):
    treq.add_last(i)

print(treq)

treq.add(5, "x")
print(treq)
```

## Exercise 2.10. Further improved RootishArrayStack


Design and implement a version of a RootishArrayStack that: 
- has only O(√n) wasted space, but 
- that can perform add(i, x) and remove(i, x) operations  

        in O(1 + min{√n, n − i}) time. 

(For an idea on how to do this, see Section 3.3.)

```python
from ods.arrays import FurtherImprovedRAS

# Arrastacks holds (bounded) ArrayDeques instead of lists
# when adding, does chained calls:
#   last_deque.add_left(second_last_deque.pop_right)
# in a loop and work backwards.
# or vice versa, when removing.

fi_ras = FurtherImprovedRAS()
for i in range(100):
    fi_ras.append(str(i))

fi_ras.add(0, "x")
fi_ras.add(50, "x")
print(fi_ras)
```

```python
# as suggested by Morin, we have one extra logical block saved with only Nones
# instead of deleting it and making it as a reserved
# --> this is faster than creating a new deque with _grow() when adding later
# but at the same time a break from the rule of wasted space.

fi_ras.remove(0)
print(fi_ras)


for i in range(11):
    fi_ras.remove(50)

print(fi_ras)
```
