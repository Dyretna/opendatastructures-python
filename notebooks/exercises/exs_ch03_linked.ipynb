## Exercise 3.17. write SEList proof

Prove that, if an SEList is used like a Stack (so that the only modifications to the SEList are done using push(x) ≡ add(size(), x) and pop() ≡ remove(size() − 1)), then these operations run in constant amortized time, independent of the value of b.

---

**Proof:**

Push and pop only touch the last block. spread and gather are never called: spread requires b following full blocks (b+1 elements each), gather requires b following blocks with b−1 elements. At the last block, the next node is the dummy, so both search loops terminate immediately. No elements are ever shifted between blocks. add(n, x) goes directly to append; remove(n − 1) works only on the last block.

Each operation is then O(1): inserting into or removing from the BoundedDeque is O(1). The only non-constant candidate is node creation and deletion. A new node is created at most every b+1 pushes, and a node is removed only after all its b+1 elements have been popped. Charging 1 per push pays for both: after b+1 pushes the saved credit covers the new node, and its removal is covered by the pops that emptied it. Nothing depends on b, so push and pop are O(1) amortized.

## Exercise 3.19: XOR-swap

Explain how to use the bitwise exclusive-or operator, ˆ, to
swap the values of two int variables without using a third variable.

```python
def xor_swap(a:int, b:int):
    print(f"input: a={a}, b={b}")
    a ^= b
    print("a ^= b:", a)
    b ^= a
    print("b ^= a:", b)
    a ^= b
    print("a ^= b:", a)

    return a, b

print("result:", xor_swap(16, 8))
```
