## Exercise 2.3. Treque() 

Design and implement a Treque (triple-ended queue). This is a List implementation in which get(i) and set(i, x) run in constant time and add(i, x) and remove(i) run in time O(1 + min{i, n − i, |n/2 − i|}) .

In other words, modifications are fast if they are near either end or near the middle of the list.

```python
from ods.arrays import Treque

# end operations does not balance the blocks,
# only with indexing add(i, x) and remove(i)

treq = Treque()
for i in range(9):
    treq.add_first(i)
print(treq)

treq = Treque()
for i in range(9):
    treq.add_last(i)

print(treq)

treq.add(5, "x")
print(treq)
```

Modify the ArrayDeque implementation so that it does not
use the mod operator (which is expensive on some systems). Instead, it
should make use of the fact that, if length(a) is a power of 2, then

        k mod length(a) = k ∧ (length(a) − 1) .

(Here, ∧ is the bitwise-and operator.)

```python
from ods.arrays import NoModDeque

nmd = NoModDeque()

nmd.add_all(0, ["a", "b", "c", "d", "e", "f", "g", "h"])
print(nmd, "\n")

```

```python
for i in range(3):
    nmd.remove_first()
print(nmd)
```

## Exercise 2.9. Improved RootishArrayStack

Design and implement a version of a RootishArrayStack that: 
- has only O(√n) wasted space, but 
- that can perform add(i, x) and remove(i, x) operations  

        in O(1 + min{i, n − i}) time.

```python
from ods.arrays import ImprovedRAS

# setup backing arrays to test get and set
# before implementation of add and remove

iras = ImprovedRAS()
iras.front.add(0, ["middle"])
iras.front.add(1, [9, 8])
iras.front.add(2, [7, 6, 5])
iras.front.add(3, [4, 3, 2, "front_end"])

iras.back.add(0, ["middle"])
iras.back.add(1, [12, 13])
iras.back.add(2, [14, 15, 16])
iras.back.add(3, [17, 18, 19,"back_end"])

iras.n = 20
iras.front_n = 10
iras.back_n = 10

print(iras)
```

```python
iras.set(0, "x")
iras.set(3, "x")
iras.set(9, "x")
iras.set(10, "x")
iras.set(16, "x")
iras.set(19, "x")

print(iras)
```
