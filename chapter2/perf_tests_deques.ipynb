# Performance tests ArrayDeque versions

```python
import time

from collection import ArrayDeque, ArrayDequeNoMod, ArrayDequeOrdered
```

```python
test_sizes = [100, 500, 1000, 5000, 10000, 50000, 100_000]

print("=== ArrayDeque Performance ===")
for n in test_sizes:
    start = time.time()
    ad = ArrayDeque()
    for i in range(n):
        ad.add(i, i)
    elapsed = time.time() - start

    print(f"n={n:>6}: {elapsed:.4f}s ({elapsed/n*1000:.4f}ms per add)")


print("\n=== ArrayDequeNoMod Performance ===")
for n in test_sizes:
    start = time.time()
    adnm = ArrayDequeNoMod()
    for i in range(n):
        adnm.add(i, i)
    elapsed = time.time() - start

    print(f"n={n:>6}: {elapsed:.4f}s ({elapsed/n*1000:.4f}ms per add)")


print("\n=== ArrayDequeOrdered Performance ===")
for n in test_sizes:
    start = time.time()
    ado = ArrayDequeOrdered()
    for i in range(n):
        ado.add_right(i)
    elapsed = time.time() - start

    print(f"n={n:>6}: {elapsed:.4f}s ({elapsed/n*1000:.4f}ms per add)")
```
